```python
import pandas as pd
import numpy as np
import os
import joblib
```

```python
test_df = pd.read_csv(r'D:\Creditcard_fraud_detection\data\processed\test_features.csv')
# pd.set_option("display.max_columns",None)
pd.set_option("display.max_rows",None)
test_df.head(90)
```

```python
model_xgb = joblib.load(r"D:\Creditcard_fraud_detection\data\save model\cardguard_xgboost.pkl")

print("Model loaded successfully!")
```

```python
# Select one random transaction
random_data = test_df.iloc[[86]].copy()

# Actual fraud label
actual = random_data["is_fraud"].iloc[0]

# Remove target
X_random = random_data.drop(columns=["is_fraud"])

# Prediction
prediction = model_xgb.predict(X_random)[0]

# Fraud probability
probability = model_xgb.predict_proba(X_random)[0, 1]

print("=" * 50)
print("CARDGUARD INFERENCE")
print("=" * 50)

print(f"Fraud Probability : {probability:.2%}")

if prediction == 1:
    print("Prediction        : 🚨 FRAUD")
else:
    print("Prediction        : ✅ LEGITIMATE")

print(f"Actual Label      : {actual}")

print("=" * 50)
```
